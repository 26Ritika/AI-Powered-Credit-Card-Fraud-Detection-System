```python
def explain_fraud_transaction(transaction):

    # Prepare transaction according to selected model
    if best_model_name == "Logistic Regression":
        model_input = scaler.transform(transaction)
    else:
        model_input = transaction

    # Predict fraud probability
    probability = best_model.predict_proba(model_input)[0, 1]

    # Generate SHAP explanation
    explanation = explainer(model_input)

    print("Model:", best_model_name)
    print("Fraud probability:", round(float(probability), 4))
    print(
        "Prediction:",
        "Fraud" if probability >= best_threshold else "Legitimate"
    )

    # Display explanation
    shap.plots.waterfall(explanation[0])

    return probability, explanation
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix

thresholds = np.linspace(0.01, 0.99, 99)

cost_results = []

for threshold in thresholds:

    predictions = (
        calibrated_val_prob >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val,
        predictions,
        labels=[0, 1]
    ).ravel()

    total_cost = (
        fn * FALSE_NEGATIVE_COST
        + fp * FALSE_POSITIVE_COST
    )

    cost_results.append({
        "Threshold": threshold,
        "False Positives": fp,
        "False Negatives": fn,
        "Total Business Cost": total_cost
    })

cost_df = pd.DataFrame(cost_results)

best_cost_row = cost_df.loc[
    cost_df["Total Business Cost"].idxmin()
]

business_threshold = float(best_cost_row["Threshold"])

print("Best Business Threshold:", business_threshold)
print("Minimum Estimated Cost:", best_cost_row["Total Business Cost"])
print("False Positives:", best_cost_row["False Positives"])
print("False Negatives:", best_cost_row["False Negatives"])
```
